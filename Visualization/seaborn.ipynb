# Seaborn

```python
import seaborn as sns
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
sns.get_dataset_names()
```

```python
df = sns.load_dataset('penguins')
df.head()
```

```python
df.shape
```

```python
df.info()
```

```python
df['species'].value_counts()
```

```python
df['island'].value_counts()
```

Scatter plot  
2 numerical variables ke beech relationship dekhne ke liye 

```python
# hue = alag category ko alag visual group/color dena.
# color = palette = "Set1"  / "Set2"  / "viridis"  /"coolwarm"

sns.scatterplot(data=df,x="bill_length_mm",y="body_mass_g", hue="species", palette="Set2")   
sns.set_style('darkgrid')               # whitegrid / white             
plt.show()
```

<h1>Strip Plot OR Swarm Plot

Strip Plot categorical + numerical data dekhne ke liye.

```python
sns.stripplot(data=df,x='species',y="body_mass_g")
```


Swarm plot dots ko aise arrange karta hai ki observations ko clearly dekha ja sake.

```python
sns.swarmplot(
    data=df,
    x="species",
    y="body_mass_g"
)
```

Difference

Strip plot

→ dots spread/random jitter ke saath.

Swarm plot

→ dots ko intelligently arrange karta hai taaki overlap kam ho.

<h1>Histogram

```python
sns.histplot(
    data=df,x="body_mass_g",bins=20,hue="species")
```

<h1>REGRESSION PLOT

Regression plot scatter + best-fit/regression line dikhata hai.

```python
sns.regplot(data=df,x="bill_length_mm",y="body_mass_g")
```

<h1>line plot

Trend dekhna, especially ordered/time-based data mein.

```python
sns.lineplot(data=df,x="bill_length_mm",y="body_mass_g")
```

<h1>Joint plot

Joint plot ek saath:  
  
main relationship  
individual distributions  
dikhata hai 

```python
sns.jointplot(data=df,x="bill_length_mm",y="body_mass_g")
```

<h1>Bar plot

Bar plot ka use category comparison ke liye.

```python
sns.barplot(data=df,x="species",y="body_mass_g")
```

<h1>COUNT PLOT

```python
sns.countplot(data=df,x="species")
df['species'].value_counts()
```

<h1> box Plot

Box plot se hum dekhte hain

data distribution  
median  
spread  
possible outliers  

```python
sns.boxplot(data=df,x="species",y="body_mass_g")
```

<h1>VIOLIN PLOT

Violin plot distribution ka shape dikhata hai.

```python
sns.violinplot(data=df,x="species",y='body_mass_g')
```

<h1>Kde plot

Numerical data ka smooth distribution curve.

```python
sns.kdeplot(data=df,x='body_mass_g',hue="species")
```

<h1>Heatmap

Heatmap data ko colors ke through represent karta hai.

```python
corr = df.corr(numeric_only=True)
sns.heatmap(corr,annot=True)
```

<h1>RUG PLOT

```python
sns.rugplot(data=df,x='body_mass_g')
```

<h1>pair plot

Ye multiple numerical columns ke pairwise relationships ek saath dikhata hai.

```python
sns.pairplot(data=df,hue='species')
```

<h1> PAIRGRID

PairGrid more customizable version hai.

```python
g = sns.PairGrid(df)
g.map_diag(sns.histplot)
g.map_offdiag(sns.scatterplot)
```
