*args

```python
def mul(*x):
    total = 1

    for a in x:
        total *= a

    return total

print(mul(10,20))
```

**kwargs

```python
def student(**details):
    return details

print(student(name="Adarsh",age=20))
```

```python
def student(**details):
    print(details["name"])
    print(details["course"])

print(student(name="Adarsh",age=20,course="Ai/Ml"))
```

Recursion

```python
def countdown(n):
    if n == 0:
        return
    print(n)
    countdown(n-1)


countdown(int(input()))
```

```python
def factorial(n):
    if n==1:
        return 1
    return n*factorial(n-1)

factorial(int(input()))
```

```python
def test(n):
    if n == 0:
        return

    print(n)
    test(n - 1)

test(3)
```

```python
def fun(n):
    if n == 1:
        return 1

    return n + fun(n - 1)

print(fun(4))
```

recursion se 1 se 5 tak print karne ka function banao.

```python
def num(n):
    if n==0:
        return
    num(n-1)
    print(n)
num(5)
```

```python
def num(n):
    if n==0:
        return 1
    print(n)
    num(n-1)
num(5)
```

```python
def sum_n(n):
    if n==1:
        return 1
    return n+sum_n(n-1)

sum_n(5)
```

```python
def fact(n):
    if n ==1:
        return 1
    return n*fact(n-1)
fact(5)
```
