```python
print(data3['Name'])
print()
print(data3[["Name","City"]])
```

```python
# Q1. Find the students who are from Delhi and have marks greater than 80.

print(data3[(data3["City"]== "Delhi") & (data3["Marks"]> 80)])
```

```python
# Q2. Find the Python course students whose age is greater than 20 and marks are greater than 70.

print(data3[(data3['Course']=="Python") & (data3['Age']>20) & (data3['Marks']>70)])
```

```python
# Q3. Find the students from Delhi or Mumbai whose marks are greater than 75.

print(data3[((data3['City']=="Delhi") | (data3['City']=="Mumbai")) & (data3['Marks']> 75)])
```

```python
# Q4. Find the ML course students whose age is greater than 20 or marks are greater than 85.

print(data3[(data3["Course"]=='ML') & ((data3['Age']>20) | (data3['Marks']>85))])
```

```python
# Q5. Find the students from Pune whose age is greater than 21 and marks are greater than 60.

print(data3[(data3['City']=='Pune') & (data3['Age']>21) & (data3['Marks']>60)])
```

```python
# Q6. Find the students who are from Delhi or Pune and whose marks are between 70 and 90.

print(data3[(((data3['City']=='Delhi')|(data3['City']=='Pune')) & (data3['Marks']>=70) & (data3["Marks"]<=90))])
```

# Handling Missing Data  
 NaN (Not a number)  
 None (for object data types)



```python
#Data

emp_data = {
    "Name": ["Aman", None, "Priya", "Neha", "Rohit","Sneha", "Vikas", "Anjali", "Karan", "Adarsh"],
    "Age": [25, 30, 28, 24, 35, 27, None, 29, 26, 23],
    "Department": ["IT", "HR", "IT", "Finance", "Sales","IT", None, "Finance", "Sales", "IT"],
    "Salary": [55000, 60000, 75000, 48000, None,70000, 65000, 58000, 52000, 90000],
    "Experience": [2, 5, 4, 1, 8, 3, 6, 4, 2, 1]
}
emp_data = pd.DataFrame(emp_data)
```

```python
# isnull()        True - NaN    , False - Value is present

print(emp_data.isnull())

#isnull().sum

print(emp_data.isnull().sum())
```

```python
#  dropna()

emp_data.dropna(inplace=True)
print(emp_data)
```

```python
# fillna()
# fillna(value, inplace=True)
emp_data['Salary'] = emp_data['Salary'].fillna(emp_data['Salary'].mean()).round(2)
emp_data['Age'] =emp_data['Age'].fillna(emp_data['Age'].mean()).astype(int)
print(emp_data)
```

```python
# interpolation
#  linear , polynomial ,time ,spline

count = {"Count" : [10,20,30,40,None,60,70]}
count = pd.DataFrame(count)

count = count.interpolate(method='linear')

print(count)
```

```python
employee_data = {
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Rohit","Sneha", "Vikas", "Anjali", "Karan", "Adarsh"],
    "Age": [25, 30, 28, 24, 35, 27, 27, 29, 25, 23],
    "Department": ["IT", "HR", "IT", "Finance", "Sales","IT", "HR", "Finance", "Sales", "IT"],
    "Salary": [55000, 60000, 75000, 48000, 82000,70000, 65000, 58000, 52000, 90000],
    "Experience": [2, 5, 4, 1, 8, 3, 6, 4, 2, 1]
}

employee_data = pd.DataFrame(employee_data)
print(employee_data)
```

merging & joining

```python
df = pd.DataFrame({
    "ID": [1, 2, 3, 4, 5],
    "Name": ["Aman", "Rahul", "Priya", "Neha", "Adarsh"],
    "Department": ["IT", "HR", "IT", "Finance", "IT"]
})

print(df)

df_salary = pd.DataFrame({
    "ID": [1, 2, 3, 6, 7],
    "Salary": [55000, 60000, 75000, 65000, 70000]
})

print(df_salary)
```

Inner Merge → Sirf woh rows jo dono DataFrames mein common hain.  
Left Merge → Left DataFrame ki saari rows + right ki matching rows.  
Right Merge → Right DataFrame ki saari rows + left ki matching rows.  
Outer Merge → Dono DataFrames ki saari rows, match ho ya na ho aur unko NaN se fill kar dega   
Cross Join → Left DataFrame ki har row ko Right DataFrame ki har row ke saath combine karta hai.

```python

print("inner")
df_merge = pd.merge(df,df_salary, on="ID", how='inner')
print(df_merge)

print("outer")
df_merge = pd.merge(df,df_salary, on="ID", how='outer')
print(df_merge)

print("left")
df_merge = pd.merge(df,df_salary, on="ID", how='left')
print(df_merge)

print("right")
df_merge = pd.merge(df,df_salary, on="ID", how='right')
print(df_merge)

print("cross")
df_merge = pd.merge(df,df_salary, how='cross')
print(df_merge)
```

concatenation

vertically(row-wise)  
horizontally(column-wise)

```python
print("vertically")
concat = pd.concat([df,df_salary])
print(concat)

print("horizontally")
concat = pd.concat([df,df_salary],axis=1)
print(concat)
```
